#### Question 7: ATM Menu

```python
balance=50000
menu=("1. Check Balance",'2. Deposit','3. Withdraw','4. Exit')
while True:
    print ("ATM MENU")
    print("Enter Number To Continue")
    print (menu)
    number=int(input("Please Enter Number"))
    if number == 1:
        print (f'Your Current Balance is : {balance}')
    elif number == 2:
        deposit=int(input("Enter Amount You Want To D3eposit "))
        balance+=deposit
        print(f'Your New Balance is : {balance}')
    elif number == 3:
        withdraw=int(input('Enter Amount You Want To Withdraw'))
        if withdraw > balance:
            print("Insufficient funds.")
        elif withdraw <=0:
            print ("invalid amount")
        else:
            balance -= withdraw
            print (f"Your Remaining Balance is : {balance}")
    elif number == 4 :
        print ('Thank You For Using ATM')
        break
    else:
        print("Invalid Choice.Please Enter Number Between 1 to 4") 
```
